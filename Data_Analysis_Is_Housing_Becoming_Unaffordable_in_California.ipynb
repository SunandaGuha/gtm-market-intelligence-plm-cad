This analysis explores whether rising housing costs in California are outpacing income — and what that means for affordability across different regions.

**Importing Python libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
```

**I selected California Housing Dataset from Kaggle https://www.kaggle.com/datasets/dhirajnirne/california-housing-data?resource=download**

```python
df = pd.read_csv('/content/sample_data/housing.csv')
df.head()
```

**Cleaning Data**

```python
df = df.dropna()
df.info()
```

**Creating Income vs House Value chart**

```python
plt.figure()
plt.scatter(df['median_income'], df['median_house_value'])
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.title('Income vs Housing Prices in California')
plt.show()
```

**Note:**

A noticeable clustering of house values at $500,000 suggests a cap in the dataset, indicating that actual housing prices may be higher than reported. This implies affordability challenges could be even more severe than observed.

```python
df['price_to_income'] = df['median_house_value'] / df['median_income']

df['price_to_income'].mean()
```

**Note:**

On average, the price-to-income ratio indicates that housing costs are disproportionately high compared to earnings, reinforcing concerns about affordability. The average price-to-income ratio further reinforces the gap between earnings and housing costs.

**Distribution of house prices**

```python
plt.figure()
plt.hist(df['median_house_value'], bins=50)
plt.xlabel('House Value')
plt.ylabel('Frequency')
plt.title('Most Homes Are Clustered at High Prices')
plt.show()
```

**Note:**

A clear horizontal ceiling at $500,000 suggests that housing prices in the dataset are capped. This indicates that actual housing costs in California may be significantly higher, meaning the affordability crisis could be even more severe than the data suggests.

**Ocean proximity comparison**

```python
df.groupby('ocean_proximity')['median_house_value'].mean().plot(kind='bar')
plt.title('Coastal Areas Are Significantly More Expensive')
plt.ylabel('Average Price')
plt.show()
```

**Note:**

Coastal regions show significantly higher housing prices compared to inland areas, highlighting geographic inequality in affordability.

Is Housing Becoming Unaffordable in California? A Data Analysis

**Introduction**

California has long faced a housing affordability crisis. This analysis explores the relationship between income and housing prices using publicly available data to understand whether rising costs are outpacing earnings.

 **Key Findings**

**1. Income vs Housing Prices**
Higher income is associated with higher house values, but the increase in housing prices appears disproportionately large compared to income levels, suggesting affordability challenges.

**2. Housing Price Distribution**

The distribution shows a concentration of high-value homes, indicating that a large portion of properties may be out of reach for average-income households.

**3. Location Matters**

Housing near the ocean is significantly more expensive than inland areas, highlighting geographic inequality in housing affordability.

**Conclusion**

The data suggests that housing affordability is a growing concern in California. Rising home prices, especially in coastal regions, are not matched proportionally by income levels, making it increasingly difficult for middle-income families to afford housing.This type of analysis can help journalists identify regions where housing inequality is most severe and guide deeper reporting.
